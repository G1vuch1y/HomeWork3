Чтение с файла для задачи А

```python
import csv
import sys

items = {}          
all_categories = set()
with open('items_A.csv', 'r', encoding='utf-8') as f:
    reader = csv.reader(f)
    header = next(reader) 
    for row in reader:
        if not row: 
            continue
        item_id = int(row[0])
        category = row[1]
        price = int(row[2])
        rating = float(row[3])
        brand = row[4]
        in_stock = int(row[5])
        
        items[item_id] = {
            'category': category,
            'price': price,
            'rating': rating,
            'brand': brand,
            'in_stock': in_stock
        }
        all_categories.add(category)
```

A1

```python
ans1 = 0
for item in items.values():
    if item['category'] == 'phones' and item['rating'] >= 4.5 and item['in_stock'] == 1:
        ans1 += 1
print("Ответ A1:", ans1)
```

А2

```python
laptop_brands = {}  
for item in items.values():
    if item['category'] == 'laptops':
        b = item['brand']
        if b not in laptop_brands:
            laptop_brands[b] = [0, 0]
        laptop_brands[b][0] += item['price']
        laptop_brands[b][1] += 1
        
max_avg = -1
best_brand = ""
for brand, (total_price, count) in laptop_brands.items():
    avg = total_price / count
    if avg > max_avg:
        max_avg = avg
        best_brand = brand
        
print("Ответ A2:", best_brand)
```

А3

```python
ans3 = 0
for item in items.values():
    if item['in_stock'] == 1:
        if item['rating'] >= 4.5 and item['price'] >= 50000:
            ans3 += 1
print("Ответ A3:", ans3)
```

ЧТЕНИЕ ГРАФА И ПОДГОТОВКА ДЛЯ A4 / A5

```python
adj = {item_id: [] for item_id in items.keys()}
viewed_to_set = set() 

with open('also_viewed_A.csv', 'r', encoding='utf-8') as f:
    reader = csv.reader(f)
    header = next(reader)
    for row in reader:
        if not row: 
            continue
        u = int(row[0])
        v = int(row[1])
        
        viewed_to_set.add(v)
        
        if u not in adj: adj[u] = []
        if v not in adj: adj[v] = []
        
        adj[u].append(v)
        adj[v].append(u)
```

A4

```python
cat_counts = {cat: 0 for cat in all_categories}
    
for item_id in viewed_to_set:
    if item_id in items:
        cat = items[item_id]['category']
        cat_counts[cat] += 1
    
sorted_cats = sorted(cat_counts.keys())
with open('answer4.csv', 'w', encoding='utf-8') as f:
    f.write("category,cnt\n")
    for cat in sorted_cats:
        f.write(f"{cat},{cat_counts[cat]}\n")
        
print("Ответ A4: Файл answer4.csv успешно сгенерирован!")
```

A5

```python
used = {}
flag_phones = False
flag_accessories = False

def dfs(u):
    global flag_phones, flag_accessories
    
    used[u] = True
 
    if u in items:
        if items[u]['category'] == 'accessories':
            flag_accessories = True
        if items[u]['category'] == 'phones':
            flag_phones = True
            
    for to in adj[u]:
        if not used[to]:
            dfs(to)

ans = 0

for i in adj.keys():
    used[i] = False
    
for i in adj.keys():
    if not used[i]:
        flag_phones = False
        flag_accessories = False
        
        dfs(i)
        
        if flag_phones and flag_accessories:
            ans += 1
            
print("Ответ A5:", ans)
```

Задача С

```python
nums = [0.0] * 10

for i in range(1, 6):
    nums[i] = nums[i-1] + 0.01

for i in range(6, 10):
    nums[i] = min(1.0, nums[i-1] + 0.25)

avg = sum(nums) / 10.0

med = (nums[4] + nums[5]) / 2.0

res = abs(avg - med)

print(f"Ответ на задачу C: {res:.6f}")
```

Задача D

```python
points = [
    (0, 0), (0, 2), (0, 7), (0, 8),
    (4, 4), (4, 7),
    (6, 1), (6, 5),
    (9, 0), (9, 1), (9, 4), (9, 7),
    (11, 2), (11, 6)
]

def get_mae(a, b):
    error = 0
    for x, y in points:
        y_pred = a * x + b
        error += abs(y_pred - y)
    return error / len(points)

min_mae = float('inf')
best_a, best_b = 0, 0

for a_int in range(-20, 21):
    a = a_int / 10.0
    for b_int in range(-20, 81):
        b = b_int / 10.0
        
        mae = get_mae(a, b)
        if mae < min_mae:
            min_mae = mae
            best_a = a
            best_b = b

min_mae_fine = float('inf')

a_start = int((best_a - 0.2) * 10000)
a_end   = int((best_a + 0.2) * 10000)

b_start = int((best_b - 0.2) * 10000)
b_end   = int((best_b + 0.2) * 10000)

for a_int in range(a_start, a_end):
    a = a_int / 10000.0
    for b_int in range(b_start, b_end):
        b = b_int / 10000.0
        mae = get_mae(a, b)
        if mae < min_mae_fine:
            min_mae_fine = mae

print(f"Ответ на задачу D: {min_mae_fine:.6f}")
```

Задача F

```python
import random

def sim():
    curr = 'S'  
    steps = 0
    while True:
        if curr == 'A':
            nxt = random.choice(['M', 'D'])
        elif curr in ('E', 'I'):
            nxt = random.choice(['T', 'S', 'M'])
        elif curr in ('T', 'M'):
            nxt = random.choice(['L', 'N'])
        elif curr in ('N', 'D'):
            nxt = random.choice(['A', 'I'])
        elif curr in ('S', 'L'):
            nxt = random.choice(['E', 'I', 'D'])
        
        steps += 1
        if curr == 'M' and nxt == 'L':
            return steps
        curr = nxt

N = 2_000_000
total = sum(sim() for _ in range(N))
print(f"Ответ на задачу F: {total / N:.2f}")
```

```python
out1 = [60, 40, 100]
out2 = [200, 50, 80]

# (-7, 6) = 3 * (-2, 3) - 1 * (1, 3)
a = 3
b = -1

ans = []

for i in range(3):
    val = a * out1[i] + b * out2[i]
    if val < 0:
        val = 0
    elif val > 255:
        val = 255
    ans.append(val)

print("Ответ на задачу B:", *ans)
```
